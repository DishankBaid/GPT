```python
with open('input.txt', 'r', encoding='utf-8') as f:
    text = f.read()
```

```python
print("length of the dataset in characters: ", len(text))
```

```python
print(text[:1000])
```

```python
chars = sorted(list(set(text)))
vocab_size =len(chars)

print("All unique chars: " , ''.join(chars))
print("vocab size: ", vocab_size)
```

```python
# Now let's do encoding and decoding part (will keep it simple char to int and back)

stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i : ch for i, ch in enumerate(chars)}

encode = lambda s : [stoi[c] for c in s] # encoder: take a string output a list of integers
decode = lambda l : ''.join(itos[i] for i in l) # decoder : takes a list of integer and output a string

print(encode("hii there"))
print(decode(encode("hii there")))
```

```python
import torch
data = torch.tensor(encode(text), dtype=torch.long) # dtype long means store them as 64 bit integers 
print(data.shape, data.dtype)
print(data[:1000]) # 1000 character which we looked at earlier will look like this to the gpt
```

```python
# Let's now split up data into train and validation sets (so that we can train against some data and validate against some)

n = int(0.9 * len(data)) # 90 % partition
train_data = data[:n] # 90% training data , rest validation
val_data = data[n:] # 10% validation data
```

```python
# Now remember we don't give everything at once to the transformer for training we give data chunks at a time.

block_size = 8
train_data[:block_size+1]
```

```python
x = train_data[:block_size] #  0 - 8 
y = train_data[1:block_size+1] # 1- 9 increament by 1 

for t in range(block_size):
    context = x[:t+1]
    target = y[t]
    print(f"when input is {context} the target: {target}")
    
```

```python
#what batch function does : At same time we choose random {batch_size} starting points 
# and then form stack of the context and target from those points.

torch.manual_seed(1337)  # just so that our random generator can reproduce same random numbers as our learning project
batch_size = 4 # how many batches to be processed parallely 
block_size = 8 # what is the maximum context length for predictions

def get_batch(split):
    # generate a small batch of data of inputs x and targets y
    data = train_data if split == 'train' else val_data
    ix = torch.randint(len(data) - block_size, (batch_size, )) # this gives you {batch_size} random numbers from 0 to data_length - block_size
    x = torch.stack([data[i : i + block_size] for i in ix]) # 
    y = torch.stack([data[i + 1 : i + block_size + 1] for i in ix])
    return x, y

xb, yb = get_batch('train')
print('inputs:')
print(xb.shape)
print(xb)

print('targets:')
print(yb.shape)
print(yb)

print('----')

# basically kinda looping through our stack of 4 * 8
for b in range(batch_size): # batch dimension
    for t in range(block_size): # time dimension 
        context = xb[b, :t+1]
        target = yb[b, t]
        print(f"When input is {context.tolist()} the target: {target}")
```

```python
print(xb) # our input to the transformer 
```

```python
import torch
import torch.nn as nn
from torch.nn import functional as F

torch.manual_seed(1337)

class BigramLanguageModel(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.token_embedding_table = nn.Embedding(vocab_size, vocab_size)

    def forward(self, idx, targets):
        logits = self.token_embedding_table(idx) 
        B, T, C = logits.shape
        logits = logits.view(B*T, C)
        targets = targets.view(B*T)
        
        loss = F.cross_entropy(logits, targets) 
        
        return logits, loss # (B, T, C)

    def generate(self, idx, max_new_tokens):
        for _ in range(max_new_tokens):
            logits, loss = self(idx)
            logits = logits[:, -1, :]
            probs = F.softmax(logits, dim = -1)
            idx_next = torch.multinomial(probs, num_samples=1)
            idx = torch.cat((idx, idx_next), dim = 1)
        return idx
        
m = BigramLanguageModel(vocab_size)
logits, loss = m(xb, yb)
print(logits.shape)
print(loss)
```
